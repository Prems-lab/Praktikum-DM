## 1. Persiapan Library dan Dataset
Menyiapkan library yang dibutuhkan serta mengimport file dataset dan melihat informasi dasar seperti head(), shape(), info(), describe()

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score
from imblearn.over_sampling import SMOTE
```

```python
df = pd.read_csv('creditcard.csv')
display(df.head())
df.shape
```

## 2. EDA dan Praproses Data
Mengidentifikasi distribusi pada atribut class, penanganan nilai duplikat, split dataset, serta menerapkan metode oversampling SMOTE

```python
class_counts = df['Class'].value_counts()
class_percentages = df['Class'].value_counts(normalize=True) * 100

print("Distribusi Kelas (Jumlah):")
print(class_counts)
print("\nDistribusi Kelas (Persentase):")
print(class_percentages)
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Class')
plt.title('Distribusi Kelas Transaksi (0 = Normal, 1 = Fraud)')

plt.yscale('log') 
plt.xlabel('Class (0: Normal, 1: Fraud)')
plt.ylabel('Count (Log Scale)')
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 4))

sns.histplot(df['Time'], bins=50, kde=True, ax=ax[0], color='blue')
ax[0].set_title('Distribusi Waktu Transaksi (Time)')
ax[0].set_xlim([min(df['Time']), max(df['Time'])])

sns.histplot(df['Amount'], bins=50, kde=True, ax=ax[1], color='red')
ax[1].set_title('Distribusi Jumlah Transaksi (Amount)')
ax[1].set_xlim([min(df['Amount']), max(df['Amount'])])
plt.show()
```

```python
jumlah_duplikat = df.duplicated().sum()
print(f"Jumlah baris duplikat: {jumlah_duplikat}")

if jumlah_duplikat > 0:
    df = df.drop_duplicates()
    print("Shape dataset setelah menghapus duplikat:", df.shape)
```

```python
X = df.drop('Class', axis=1)
y = df['Class']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

print("Distribusi kelas pada data latih SEBELUM SMOTE:")
print(y_train.value_counts())
```

```python
smote = SMOTE(random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(X_train, y_train)

print("\nDistribusi kelas pada data latih SETELAH SMOTE:")
print(y_train_smote.value_counts())
```

## 3. Membangun Model Klasisfikasi
Mempersiapkan fungsi evaluasi model dan memmbangun model menggunakan random forest yang kemudian dibandingkan antara model yang menggunakan SMOTE dan yang tidak

```python
def evaluate_model(model, X_test, y_test, model_name="Model"):
    y_pred = model.predict(X_test)
    y_pred_proba = model.predict_proba(X_test)[:, 1]
    
    print(f"========== Evaluasi {model_name} ==========")
    
    print("Classification Report:")
    print(classification_report(y_test, y_pred))
    
    auc_roc = roc_auc_score(y_test, y_pred_proba)
    print(f"AUC-ROC Score: {auc_roc:.4f}\n")
    
    print("Confusion Matrix:")
    cm = confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False)
    plt.title(f'Confusion Matrix - {model_name}')
    plt.ylabel('Actual Label')
    plt.xlabel('Predicted Label')
    plt.show()
```

```python
rf_base = RandomForestClassifier(random_state=42, n_jobs=-1)
rf_base.fit(X_train, y_train)

evaluate_model(rf_base, X_test, y_test, model_name="Random Forest (Base - Tanpa SMOTE)")
```

```python
rf_smote = RandomForestClassifier(random_state=42, max_depth=15, n_jobs=-1)
rf_smote.fit(X_train_smote, y_train_smote)

evaluate_model(rf_smote, X_test, y_test, model_name="Random Forest (Dengan SMOTE)")
```
